```python
import networkx as nx
import pathlib
```

```python
files = list(pathlib.Path('drive-download-20230913T033307Z-001').glob('*.graphml*'))
```

```python
mask_transitions_fname = [f for f in files if "mask" in f.name][0]
crop_transitions_fname = [f for f in files if "crop" in f.name][0]
```

```python
mask_g = nx.read_graphml(mask_transitions_fname)
crop_g = nx.read_graphml(crop_transitions_fname)
mask_g.add_nodes_from(crop_g.nodes(data=True))
crop_g.add_nodes_from(mask_g.nodes(data=True))
```

```python
layout = nx.circular_layout(mask_g)
```

```python
import math
```

```python
import matplotlib.pyplot as plt
# SVG text:
plt.rcParams['svg.fonttype'] = 'none'
plt.figure(figsize=(6, 6), dpi=150)

# Remove zero-weight edges:
mask_g.remove_edges_from([(u, v) for u, v, d in mask_g.edges(data=True) if d["weight"] == 0])
nx.draw_networkx_nodes(mask_g, pos=layout, node_size=5000, node_color="#00babe")
nx.draw_networkx_labels(mask_g, pos=layout,)
weights = [math.log(1+d["weight"]) for _, _, d in mask_g.edges(data=True)]
arrows = nx.draw_networkx_edges(
    mask_g,
    pos=layout,
    node_size=5000,
    # Scale to weight:
    width=weights,
    # Curved edges:
    connectionstyle="arc3,rad=0.1",
    # Draw arrows using "sharp" rendering from matplotlib:
    arrowsize=10,
        arrowstyle='-|>'  # I personally think this style scales better
)

for a, w in zip(arrows, weights):
    
    # mutation_scale affects only the arrowhead size, not the arrow tail.
    # The constants here are arbitrary; you may want/need to change them
    a.set_mutation_scale(20 + w)
    
    # Sharpen arrowheads by creating a mitered joint between arrowhead 
    # line segments, instead of the default joinstyle='round'
    a.set_joinstyle('miter')
    
    # Prevent each arrow tail from jutting forward and through the arrowhead,
    # which happens with the default capstyle='projecting'
    a.set_capstyle('butt')

plt.savefig("mask-transitions.svg")
```
