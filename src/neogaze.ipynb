```python
# Imports
from PIL import Image
import numpy as np
import sys
import cv2
import os
```

```python
# Function to find the green dot:
def subset_image_around_centroid(image, centroid, radius):
    """
    Returns a subset of the image centered on the centroid.
    """
    x, y = centroid
    x_min = max(0, x - radius)
    x_max = min(image.shape[0], x + radius)
    y_min = max(0, y - radius)
    y_max = min(image.shape[1], y + radius)
    return image[x_min:x_max, y_min:y_max]
```
